**REGRESION LINEAL**

```python
X=df[['Daily Obs Count','Site Latitude','Site Longitude']]

y=df['Daily Max 8-hour CO Concentration']

X.head()
```

```python
X.corr().round(3)
```

```python
y.head()
```

```python
X.isna().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=123)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
lm=LinearRegression()
```

```python
lm.fit(X_train,y_train)
```

```python
print("El termino de intersección del modelo líneal:",lm.intercept_)
```

```python
print("Los coeficientes del modelo líneal:",lm.coef_)
```

```python
cdf=pd.DataFrame(data=lm.coef_,index=X_train.columns,columns=["coeficientes"])

cdf
```

```python
from matplotlib import gridspec

l=list(cdf.index)

l
```

```python
fig=plt.figure(figsize=(10,7))

gs=gridspec.GridSpec(1,3)

titulos=['N.º de observaciones','Latitud','Longitud']

for i in range(3):

    ax=plt.subplot(gs[i])

    ax.scatter(df[l[i]],df['Daily Max 8-hour CO Concentration'])

    ax.set_title(titulos[i],fontdict={'fontsize':16})

plt.tight_layout()
plt.show()
```

```python
n=X_train.shape[0]

k=X_train.shape[1]

dfN=n-k

train_pred=lm.predict(X_train)

train_error=np.square(train_pred-y_train)

sum_error=np.sum(train_error)

se=[0.0]*k

for i in range(k):

    mse=sum_error/dfN
    variance_coeff=mse/np.sum(np.square(X_train[X_train.columns[i]]-X_train[X_train.columns[i]].mean()))

    se[i]=np.sqrt(variance_coeff)

cdf['Standard Error']=se

cdf['t-statistic']=(cdf['coeficientes']/cdf['Standard Error'])

cdf
```

```python
predictions=lm.predict(X_test)

print("Tipo del objeto predicho:",type(predictions))

print("Tamaño del objeto predicho:",predictions.shape)
```

```python
print("MAE:",metrics.mean_absolute_error(y_test,predictions))

print("MSE:",metrics.mean_squared_error(y_test,predictions))

print("RMSE:",np.sqrt(metrics.mean_squared_error(y_test,predictions)))

print("R²:",metrics.r2_score(y_test,predictions))
```

```python
plt.figure(figsize=(10,7))

plt.title("Concentración de CO: real vs. predicha",fontsize=20)

plt.xlabel("Concentración de CO real",fontsize=16)

plt.ylabel("Concentración de CO predicha",fontsize=16)

plt.scatter(x=y_test,y=predictions)

plt.show()
```

```python
plt.figure(figsize=(10,7))

plt.title("Histograma de residuos",fontsize=20)

plt.xlabel("Residuos",fontsize=16)

plt.ylabel("Densidad",fontsize=16)
sns.distplot(y_test-predictions)

plt.show()
```

```python
plt.figure(figsize=(10,7))

plt.title("Residuos vs. valores predichos",fontsize=20)

plt.xlabel("Concentración de CO predicha",fontsize=16)

plt.ylabel("Residuos",fontsize=16)

plt.scatter(x=predictions,y=y_test-predictions)

plt.show()
```

**RANDOM FOREST**

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf=RandomForestRegressor(n_estimators=100,random_state=123)
```

```python
rf.fit(X_train,y_train)
```

```python
predictions_rf=rf.predict(X_test)

print("Tipo del objeto predicho:",type(predictions_rf))

print("Tamaño del objeto predicho:",predictions_rf.shape)
```

```python
print("MAE:",metrics.mean_absolute_error(y_test,predictions_rf))

print("MSE:",metrics.mean_squared_error(y_test,predictions_rf))

print("RMSE:",np.sqrt(metrics.mean_squared_error(y_test,predictions_rf)))

print("R²:",metrics.r2_score(y_test,predictions_rf))
```

```python
plt.figure(figsize=(8,4))

plt.title("Random Forest: concentración de CO real vs. predicha",fontsize=20)

plt.xlabel("Concentración de CO real",fontsize=16)

plt.ylabel("Concentración de CO predicha",fontsize=16)

plt.scatter(x=y_test,y=predictions_rf)

plt.show()
```
